```python
import pandas as pd
import sklearn
```

```python
df = pd.read_pickle('./f1_23_monaco.pkl')
df
```

```python
data_aston = df.loc[df['Team'] == "Aston Martin"]
data_aston.loc[:, ['Class']] = 0 
data_aston.loc[data_aston['Driver'] == "ALO",['Class']] = 1
data_aston.loc[data_aston['Driver'] != "ALO",['Driver', 'Class']]
```

```python
data_aston['Sector1Time'] = data_aston['Sector1Time'].dt.total_seconds()
data_aston['Sector2Time'] = data_aston['Sector2Time'].dt.total_seconds()
data_aston['Sector3Time'] = data_aston['Sector3Time'].dt.total_seconds()
data_aston['Class'] = (data_aston['Driver'] == 'ALO').astype(int)
```

```python
X = data_aston.drop(columns='Class')
Y = data_aston['Class']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, random_state=2533)
```

```python
scaler = StandardScaler()

X_std     = scaler.fit_transform(X_train)  # fit + scale
X_test_std = scaler.transform(X_test)      # scale only!
```
